```python
import pandas as pd
import numpy as np
```

```python
Payments=spark.table("python.default.`1790098705165_payments`")
Payments=Payments.toPandas()

display(Payments)
```

```python
# Check the number of rows and columns
Payments.shape
```

```python
# Check the data types of each column
Payments.dtypes
```

```python
# Check for missing values in each column
print(Payments.isna().sum())
```

```python
# Check for duplicate rows
Payments.duplicated().sum()
```

```python
# Check for duplicate PaymentIDs
print(Payments["PaymentID"].duplicated().sum())
```

```python

# Count how many payments are in each status
print(Payments["PaymentStatus"].value_counts())
```

```python
# See all different payment statuses
print(Payments["PaymentStatus"].unique())
```

```python
# Remove extra spaces
Payments["PaymentStatus"] = Payments["PaymentStatus"].str.strip()

# Make all statuses lowercase
Payments["PaymentStatus"] = Payments["PaymentStatus"].str.lower()

# Check the cleaned values
print(Payments["PaymentStatus"].unique()) 
```

```python
# 1. Convert transaction_date into a proper date
Payments["PaymentDate"] = pd.to_datetime(Payments["PaymentDate"])

# 2. Extract the year
Payments["year"] = Payments["PaymentDate"].dt.year

# 3. Extract the month number
Payments["month"] = Payments["PaymentDate"].dt.month

# 4. Extract the day of the month
Payments["day"] = Payments["PaymentDate"].dt.day

# 5. Extract the day name
Payments["day_of_week"] = Payments["PaymentDate"].dt.day_name()

# 6. Create a month-year column
Payments["month_year"] = Payments["PaymentDate"].dt.to_period("M")

# 7. Display the result
display(Payments)
```
