```python
# duplicated() checks if the same row appears more than once.
# sum() counts the number of duplicate rows.
print(product.duplicated().sum())
print(product["ProductID"].duplicated().sum())
```

```python
# isnull() checks every cell and asks:
# "Is this cell empty/missing?"
# sum() then counts how many missing values there are in each column.
# So the result tells us how many blanks
# each column contains.

product.isnull().sum()
```

```python
# Check the numerical columns in the products table.
# The most important number here is UnitPrice.
# We want to see the cheapest and most expensive
# products and check whether those prices make sense.

product.describe()
```

```python
# describe() gives us statistics about UnitPrice.
# We are mainly interested in the minimum and maximum price.
# This helps us identify prices that may not make sense,

print(product["UnitPrice"].describe())
```

```python
# Sort the products from cheapest to most expensive.
# ascending=True means:
# smallest price → largest price

product.sort_values('UnitPrice', ascending=True)
```

```python
 # Sort the products from most expensive to cheapest.
# ascending=False means:
# largest price → smallest price

product.sort_values('UnitPrice', ascending=False)
```

```python
 #Shows the distinct values in the ProductName column
 #This helps us to see different peoduct we have in our dataset.
product["ProductName"].unique()
```

```python
print(product["UnitPrice"].min())
print(product["UnitPrice"].max())
print(product["UnitPrice"].mean())
print(product["UnitPrice"].median())
```

```python
# strip() removes unnecessary spaces before and after text.
product["ProductName"] = product["ProductName"].str.strip()
product["Category"] = product["Category"].str.strip()

print(product["Category"].unique())
```

```python
 # Find products with zero or negative prices
zero_or_negative = product[product["UnitPrice"] <= 0]
if len(zero_or_negative) == 0:
    print("No products with zero or negative prices found.")
else:
    display(zero_or_negative)
```
